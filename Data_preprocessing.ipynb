# Data Preprocessing

```python
df['Price'].unique()
```

**o/p variable** 

```python
df['Price'] = df['Price'].str.replace('₹ ','')
df['Price'] = df['Price'].str.replace(',','')
df['Price'] = df.Price.str.replace(' ','')
df['Price'] = df.Price.astype('int')
```

**i/p variable**

```python
df['Brand'] = df['Name'].str.extract(r'\d{4}\s(\w+)')
df['Model'] = df['Name'].str.extract(r'\d{4}\s\w+\s(\w+)')
```

```python
df['Name'].nunique()
```

```python
df['Rate'] = df['Rating'].str.extract(r'(\d+\.?\d*)').astype(float)
```

```python
df['Kilometers']=df['Kilometers'].str.replace(',','')
df['Kms']=df['Kilometers'].str.extract(r'(\d+\.?\d*)').astype(int)
```

```python
df['years_old']=df['Year_of_Purchase'].str.extract(r'(\d+\.?\d*)').astype(int)
```

```python
df['Fuel_Type']=df['Fuel_Type'].str.replace('+','and')
```

```python
# Extracting the 2-letter RTO code (after 'RTO' and before digits)
df['RTO_state_code'] = df['RTO'].str.extract(r'RTO([A-Z]{2})\d+')
```

```python
df['years_old']=2024-df['years_old']
```

```python
df=df.drop(columns=['Name','RTO','Insurance','Rating','Kilometers','Year_of_Purchase'],axis=1)
```

# Encoding

```python
df['Owner'].unique()
```

```python
owner_mapping = {"OwnerFirst Owner": 1, "OwnerSecond Owner": 2, "OwnerThird Owner": 3, 'OwnerFourth Owner':4, 'OwnerFifth Owner':5, 'OwnerSixth Owner': 6, 'OwnerSeventh Owner':7, 'OwnerEighth Owner': 8, 'OwnerNinth Owner': 9,'OwnerTenth Owner':10}
df["Owner"] = df["Owner"].map(owner_mapping).astype('int')
```

```python
df.Transmission.unique()
```

```python
df['Transmission'] = df.groupby('Brand')['Transmission'].transform(
    lambda x: x.fillna(x.mode()[0]) if not x.mode().empty else x.fillna('Unknown')
)
```

```python
df['Transmission'] = df['Transmission'].map({'TransmissionMANUAL': 0, 'TransmissionAUTOMATIC': 1, 'Unknown':2})
```

```python
df['Insurance_Type'] = (df['Insurance_Type'] == 'Comprehensive').astype(int)
```

```python
brand_mean_price = df.groupby('Brand')['Price'].mean().to_dict()
df['Brand'] = df['Brand'].map(brand_mean_price)
```

```python
model_mean_price = df.groupby('Model')['Price'].mean().to_dict()
df['Model'] = df['Model'].map(model_mean_price)
```

```python
df['RTO_state_code'].fillna('Unknown', inplace=True)
```

```python
rto_mean_price = df.groupby('RTO_state_code')['Price'].mean().to_dict()
df['RTO_state_code'] = df['RTO_state_code'].map(rto_mean_price)
```

```python
from sklearn.preprocessing import LabelEncoder
df['Fuel_Type'] = LabelEncoder().fit_transform(df['Fuel_Type'])
```

```python
x=df.drop('Price',axis=1)
y=df['Price']
```

```python
r2_Train=[]
cv=[]
for i in range(0,100):
    from sklearn.model_selection import train_test_split
    x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=i)
    
    from sklearn.linear_model import LinearRegression
    model=LinearRegression()
    model.fit(x_train,y_train)
    
    ypred_train=model.predict(x_train)
    
    from sklearn.metrics import r2_score
    r2_Train.append(r2_score(y_train,ypred_train))
    from sklearn.model_selection import cross_val_score
    cv.append(cross_val_score(model,x_train,y_train,cv=5).mean())
    
em=pd.DataFrame({"r2_Train":r2_Train,'cv':cv})
gm=em[(abs(em['r2_Train']-em['cv'])<0.05)] # for values
rs=gm[gm['cv']==gm['cv'].max()].index[0]
print("best random state number:",rs)
```
